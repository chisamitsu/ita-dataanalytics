### Algoritmo

#### Proceso

1. Arreglo la codificación de los nombres de las columnas y de los textos. En el ejercicio anterior vi que los caracteres mal decodificados empiezan por `Ã` (letras con acento) o aparecen como `â‚¬` (el símbolo `€`). Primero los busco y los cuento, mostrando un ejemplo de cada caso para identificar qué carácter debería ser. Con esa información creo un diccionario de correcciones y aplico `replace`. Al final compruebo que no quede ninguno.

   - Nota: Existen librerías que lo corrigen automáticamente, pero aquí uso un diccionario propio.

2. Cambio el nombre de la columna `Unnamed: 0` por `ID` y elimino los duplicados exactos (sin tener en cuenta el `ID`), conservando el registro con el `ID` más alto.

3. Reviso los DNI repetidos. Para cada DNI compruebo si las filas pertenecen a la misma persona verificando que coincidan el nombre, el primer apellido y la fecha de nacimiento. Con el primer apellido identifico también casos como *Laia Navarro* y *Laia Navarro Vila*. Si corresponde a la misma persona, me quedo con la fila que tiene el `ID` más alto. Si después de esta revisión todavía quedan DNI duplicados, asumo que pertenecen a personas distintas, por lo que mantengo las filas pero elimino el DNI.

   - Nota: No dispongo de una fecha de creación de cada fila, así que utilizo el `ID` para identificar el registro más reciente. Me parece una hipótesis razonable, ya que el salario suele aumentar siguiendo esta lógica, excepto en un caso de Oriol en la fila 754. Sin embargo, esa fila no contiene información relevante, por lo que considero que se trata de un error en los datos.

4. Elimino las filas que no pueden recuperarse: aquellas que no tienen nombre o apellidos, así como la fila con valores inventados que detecté al final del dataset en el ejercicio anterior.

#### Funciones

- `find_broken_characters(df)` busca los caracteres mal codificados en los nombres de columna y en los textos, y devuelve cuántas veces sale cada uno con un ejemplo. La uso dos veces: antes de corregir, para saber qué hay que cambiar, y después, para comprobar que no queda ninguno.

El resto de pasos se hace directamente con métodos de pandas.

### Implementación

#### Preparación

Preparo la lista de valores inventados (los de la última fila que vi en el ejercicio anterior), las columnas que uso para reconocer a una persona y los campos que no pueden faltar.

```python
# values in the last row of the dataset (exercise 1)
PLACEHOLDERS = ["???", "###", "XXXX", "Narnia", "Desconeguda", "??", "molts", "Z"]

# columns that identify a person (the first surname also matches "Navarro" with "Navarro Vila")
PERSON_COLS = ["Nom", "Primer cognom", "Dia de Naixement", "Mes de Naixement", "Any de Naixement"]

# columns shown in the DNI tables
DISPLAY_COLS = ["ID", "Nom", "Cognoms", "DNI",
       "Dia de Naixement", "Mes de Naixement", "Any de Naixement", "Gènere",
       "Salari mensual", "Grup Professional", "Càrrec"]

# without them we don't know who the person is
REQUIRED_COLS = ["Nom", "Cognoms"]
```

#### Funciones

```python
def find_broken_characters(df):
  text_values = pd.Series(df.columns)
  for col in df.select_dtypes("object").columns:
    text_values = pd.concat([text_values, df[col].dropna().astype("str")])

  broken_characters = pd.DataFrame({"value":text_values})
  broken_characters["char"] = broken_characters["value"].str.findall(r"Ã.|â‚¬")
  broken_characters = broken_characters.explode("char").dropna()
  broken_characters = broken_characters.groupby("char").agg(occurrences=("char","count"),example=("value","first"))

  return broken_characters.sort_values(by="occurrences",ascending=False)
```

### Ejecución

#### 1. Codificación

Primero busco qué caracteres están mal codificados

```python
print(find_broken_characters(df))
```

16 caracteres distintos. Con el ejemplo de cada uno, puedo identificar cuál debería ser el carácter correcto (`MuÃ±oz` → `Muñoz`, `TromsÃ¸` → `Tromsø`, etc).

Tres casos contienen un carácter oculto después de la `Ã` (`CÃ rrec`, `VÃ­ctor`, `Ãlvarez`). En estos casos utilizo su código correspondiente: `\xa0` (à), `\xad` (í) y `\x81` (Á).

```python
ENCODING_FIXES = {
    "â‚¬": "€",
    "Ã\xa0": "à",
    "Ã¨": "è",
    "Ã\xad": "í",
    "Ã¡": "á",
    "Ã§": "ç",
    "Ã±": "ñ",
    "Ã³": "ó",
    "Ã©": "é",
    "Ã\x81": "Á",
    "Ãº": "ú",
    "Ã¼": "ü",
    "Ã¶": "ö",
    "Ã‰": "É",
    "Ã¸": "ø",
    "Ã¤": "ä",
}

# fix column headers
df.columns = df.columns.to_series().replace(ENCODING_FIXES, regex=True)
# fix all values
df = df.replace(ENCODING_FIXES, regex=True)

print("Columnas:", df.columns.tolist())
print(f"Caracteres mal codificados después de corregir: {len(find_broken_characters(df))}")

df.head()
```

#### 2. Duplicados exactos

La primera columna es el identificador de la fila, así que la llamo `ID` y no la tengo en cuenta al comparar. El fichero no está ordenado por `ID` (en las últimas filas del ejercicio anterior sale el 651 entre el 1000 y el 1002), así que primero ordeno para que `keep="last"` se quede con el `ID` más alto.

```python
# Rename identifier column
df.columns = ["ID"] + df.columns[1:].tolist()

rows_before = len(df)

# Exact duplicates (same values in all columns except ID)
duplicated_rows = (
    df.sort_values(by="ID", ascending=False)
      .loc[lambda x: x.duplicated(subset=x.columns[1:],keep="first")]
)

print(f"----- Filas duplicadas exactas (a eliminar: {len(duplicated_rows)}) ".ljust(80, "-"))
display(duplicated_rows)

df = df.drop(duplicated_rows.index).reset_index(drop=True)

rows_after = len(df)

print(f"\n===== Filas iniciales: {rows_before}  Filas finales: {rows_after}  Eliminadas en total: {rows_before - rows_after}")
```

#### 3. DNI repetidos

Antes de limpiar, miro las filas con DNI repetido, ordenadas por `ID`.

```python
# DNI appearing more than once
duplicated_dni = (
    df.groupby("DNI")
      .filter(lambda x: len(x) > 1)
      .sort_values(by=["DNI", "Nom", "ID"],ascending=[True, True, False])
)

duplicated_dni_count = duplicated_dni["DNI"].nunique()
print(f"----- DNI duplicados ({duplicated_dni_count} DNI, {len(duplicated_dni)} filas) ".ljust(80, "-"))
display(duplicated_dni[DISPLAY_COLS])
```

Ordenando por `ID`, observo que el salario de cada persona aumenta o se mantiene, excepto en la fila 754 de Oriol Gómez, que tampoco tiene asignado un `Grup Professional`. Considero razonable asumir que el registro con el `ID` más alto es el más reciente.

Antes de hacerlo, compruebo si existen filas con los mismos valores en `PERSON_COLS` pero con un `DNI` diferente.

```python
# Extract first surname
df["Primer cognom"] = df["Cognoms"].str.split().str[0]

# Records with same PERSON_COLS
same_person = df[df.duplicated(subset=PERSON_COLS, keep=False)]

# Same person but different DNI
different_dni = (same_person.groupby(PERSON_COLS)["DNI"].transform("nunique").ne(1))

rows_to_review = (same_person[different_dni].sort_values(by=PERSON_COLS)[DISPLAY_COLS])

print(f"----- Datos duplicados con DNI diferentes ({len(rows_to_review)}) ".ljust(80, "-"))
display(rows_to_review)
```

Hay 12 filas con el mismo nombre, primer apellido y fecha de nacimiento pero con un DNI diferente. El género cambia (Adrià Ferrer aparece como M, H y D) y el salario sube y baja sin orden, así que parecen personas distintas o datos poco fiables.

No las elimino, pero en un caso real las revisaría con otra fuente de datos.

```python
# Remove duplicates with same DNI + PERSON_COLS, keep highest ID
same_person_rows = (
    df.sort_values(by="ID", ascending=False)
      .loc[lambda x: x.duplicated(subset=["DNI"] + PERSON_COLS,keep="first")]
)

print(f"----- Datos de personas duplicados exactos (a eliminar: {len(same_person_rows)}) ".ljust(80, "-"))
display(same_person_rows)
df = df.drop(same_person_rows.index).reset_index(drop=True)

print("\n----- DNI Duplicados con datos de personas diferentes (borrar el DNI) ".ljust(80, "-"))
shared_dni_rows = (
    df.groupby("DNI")
      .filter(lambda x: len(x) > 1)
      .sort_values(by=["DNI", "Nom", "ID"],ascending=[True, True, False])
)
display(shared_dni_rows)
df.loc[shared_dni_rows.index, "DNI"] = np.nan

print(f"DNI repetidos después de limpiar: {df['DNI'].dropna().duplicated().sum()}")

rows_after = len(df)
print(f"\n===== Filas iniciales: {rows_before}  Filas finales: {rows_after}  Eliminadas en total: {rows_before - rows_after}")

# Primer cognom is no longer needed
df = df.drop(columns="Primer cognom")
```

#### 4. Filas que no se pueden recuperar

```python
# Rows containing any placeholder value
fake_data = df.isin(PLACEHOLDERS).any(axis=1)

# Rows with missing values in required columns
null_data = df[REQUIRED_COLS].isna().any(axis=1)

rows_to_remove = df[fake_data | null_data]

print(f"----- Filas a eliminar ({len(rows_to_remove)}) ".ljust(80, "-"))
display(rows_to_remove)

df = df.drop(rows_to_remove.index).reset_index(drop=True)

rows_after = len(df)

print(f"\n===== Filas iniciales: {rows_before}  Filas finales: {rows_after}  Eliminadas en total: {rows_before - rows_after}")
```

### Interpretación de los resultados

1. He corregido la codificación de los nombres de columna y de los textos: había 16 caracteres mal codificados.

1. Después he limpiado los registros repetidos:
   - 4 filas duplicadas exactas.
   - 7 filas de personas repetidas con el mismo DNI. De cada persona mantengo el registro con el ID más alto, que considero el más reciente.
